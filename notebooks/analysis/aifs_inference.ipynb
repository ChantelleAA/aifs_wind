# Revised Roadmap: AIFS Skill for Offshore & Coastal Extreme Winds (Ireland)

**Focus**: AIFS skill for offshore and coastal extreme wind events around Ireland; compare offshore vs coastal vs onshore; assess AIFS ensemble value vs deterministic and vs IFS.

**Vantage points (point-based, no spatial averaging)**
- Onshore: Dublin, Shannon, Cork, Galway (Met Éireann obs + ERA5)
- Coastal: Valentia, Sherkin Island, west/southwest sites (obs + ERA5)
- Offshore: Atlantic, Celtic Sea, Irish Sea fixed points or buoys (ERA5 primary; obs if available)

**Data**
- Forecasts: AIFS-Single, AIFS-Ensemble, IFS (u10, v10 at consistent lead times)
- Reference: ERA5 (primary), Met Éireann obs where available

**Processing**
- Time align per point: forecast valid time ↔ ERA5 ↔ obs (UTC, 10 m, same units)
- Compute wind speed: sqrt(u^2 + v^2) for each model/member/time
- Build point time series for all models, ERA5, obs

**Extremes (local definitions)**
- Thresholds: 15–20 m/s offshore, or top 5–10% per point, or known storm windows

**Evaluation per point**
- Deterministic: bias, RMSE, correlation, missed extremes, false alarms, timing error, magnitude error
- Ensemble: exceedance probability, spread, lead-time skill; did ensemble flag risk earlier?

**Comparisons**
- By category: offshore vs coastal vs onshore
- Metrics: bias patterns, extreme detection rate, ensemble usefulness

**Outputs/interpretation**
- Offshore wind energy relevance
- Coastal hazard/early warning insight
- Operational value of AIFS ensemble

4) Define extreme thresholds (local percentiles + fixed 15–20 m/s offshore).
5) Compute metrics per point and per category.
